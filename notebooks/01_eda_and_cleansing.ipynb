```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset load
df = pd.read_csv('../data/raw_telecom_churn.csv')

# Dataset structure & first 5 rows
print(df.shape)
df.head()
```

```python
# convert blankspaces into float in total charges
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].replace(' ', np.nan), errors='coerce')

# fill missing values with medians
df['TotalCharges'].fillna(df['TotalCharges'].median(), inplace=True)

# Churn column into binary
df['Churn'] = df['Churn'].apply(lambda x: 1 if str(x).strip().lower() in ['yes', '1', 'true'] else 0)

# Check null values count
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Save cleaned dataframe
df.to_csv('../data/processed_churn_data.csv', index=False)
print("Cleaned data successfully saved to data/processed_churn_data.csv")
```
