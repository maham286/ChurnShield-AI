```python
import pandas as pd
import numpy as np

# load cleaned data
df = pd.read_csv('../data/processed_churn_data.csv')

# Behavioral Features Engineering
# Tenure Segmentation
df['TenureGroup'] = pd.cut(
    df['tenure'], 
    bins=[-1, 12, 24, 48, 72, 100], 
    labels=['0-1 Year', '1-2 Years', '2-4 Years', '4-6 Years', '6+ Years']
)

# Charge Velocity Ratio
df['ChargeRatio'] = df['TotalCharges'] / (df['MonthlyCharges'] + 1e-5)

# One-Hot Encoding (Convert Categorical variables into numbers)
cat_cols = df.select_dtypes(include=['object', 'category']).columns
if 'customerID' in cat_cols:
    cat_cols = cat_cols.drop('customerID')

df_encoded = pd.get_dummies(df, columns=cat_cols, drop_first=True)

# Save Feature Engineered Data
df_encoded.to_csv('../data/engineered_churn_data.csv', index=False)
print("Features engineered and saved to data/engineered_churn_data.csv!")
```
